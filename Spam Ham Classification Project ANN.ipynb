```python
import pandas as pd
```

```python
df = pd.read_csv('/content/SMSSpamCollection.csv', sep='\t', names = ['label', 'messages'])
```

```python
df.head()
```

```python
from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt_tab')
```

```python
corpus = df['messages']
```

```python
from nltk.tokenize import sent_tokenize
```

```python
from gensim.utils import simple_preprocess
```

```python
words = []
for sent in corpus:
  words.append(simple_preprocess(sent))
```

```python
words
```

```python
model = gensim.models.Word2Vec(words, vector_size=300 ,min_count=2, epochs=10)
```

```python
model.corpus_count
```

```python
model.epochs
```

```python
model.wv.index_to_key
```

```python
model.wv.most_similar('king')
```

```python
model.wv.most_similar('kid')
```

```python
model.wv.similarity('kid', 'child')
```

```python
import numpy as np
```

```python
def avgword2vec(doc):
  valid_words = [model.wv[word] for word in doc if word in model.wv.index_to_key]
  if len(valid_words) > 0:
    return np.mean(valid_words, axis=0)
  else:
    return np.zeros(model.vector_size)
```

```python
!pip install tqdm
```

```python
from tqdm import tqdm
```

```python
X = []
for i in tqdm(range(len(words))):
  X.append(avgword2vec(words[i]))
```

```python
len(X)
```

```python
X[0].reshape(1,-1).shape
```

```python
X_n = np.array(X)
X_n.shape
```

```python
df_= pd.DataFrame(X_n)
```

```python
df_.head()
```

```python
y = df['label'].map({'ham':1, 'spam':0})
```

```python
df_['Output'] = y
```

```python
df_.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_.drop('Output', axis=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df_.drop('Output', axis=1), y, test_size=0.25, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(n_estimators=200, criterion = 'gini')
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)
```

```python
y_pred_proba = rf.predict_proba(X_test)
y_pred_proba[:,:1]
```

```python
from sklearn.metrics import roc_auc_score, classification_report
```

```python
print(classification_report(y_test,y_pred))
print('ROC-AUC Score ',roc_auc_score(y_test, y_pred))
```

#### HYPERPARAMETER TUNING TO AVOID OVERFITTING

```python
params_grid = {
    'n_estimators': [50,100,250,500]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(estimator=RandomForestClassifier(), param_grid=params_grid, cv=5)
```

```python
grid.fit(X_train, y_train)
```

```python
y_pred = grid.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(roc_auc_score(y_test, y_pred))
```
